```python
from pathlib import Path

import pandas as pd


root_folder = Path(r"c:\Users\Michael Migacev\Desktop\UNI\Wirklich_UNI\AICON\master-thesis-tol\master-thesis-tol\tower_of_london_plots\problem_ordering_outputs\po_cv_outputs_12_12_v2_extra_number_of_moves")


def build_top5_training_csv(training_results_path: Path) -> pd.DataFrame:
    training_df = pd.read_csv(training_results_path)
    required_columns = {
        "alpha",
        "beta",
        "kendall_tau_random_tie_mean",
        "kendall_tau_random_tie_std",
    }
    missing_columns = required_columns - set(training_df.columns)
    if missing_columns:
        raise ValueError(f"{training_results_path} is missing columns: {sorted(missing_columns)}")

    top5_df = (
        training_df.sort_values(
            ["kendall_tau_random_tie_mean", "kendall_tau_random_tie_std", "alpha", "beta"],
            ascending=[False, True, True, True],
        )
        .head(5)
        .loc[:, ["alpha", "beta", "kendall_tau_random_tie_mean", "kendall_tau_random_tie_std"]]
        .rename(
            columns={
                "kendall_tau_random_tie_mean": "mean",
                "kendall_tau_random_tie_std": "std",
            }
        )
        .reset_index(drop=True)
    )

    output_path = training_results_path.with_name("top5_training.csv")
    top5_df.to_csv(output_path, index=False)
    return top5_df


def generate_top5_training_csvs(folder: Path) -> list[Path]:
    if not folder.exists():
        raise FileNotFoundError(f"Folder does not exist: {folder}")

    written_files: list[Path] = []
    split_dirs = sorted(
        path for path in folder.rglob("split_*") if path.is_dir() and path.name.startswith("split_")
    )

    for split_dir in split_dirs:
        training_files = sorted(split_dir.rglob("training_results_long.csv"))
        for training_results_path in training_files:
            build_top5_training_csv(training_results_path)
            written_files.append(training_results_path.with_name("top5_training.csv"))
            print(f"Saved {training_results_path.with_name('top5_training.csv')}")

    if not written_files:
        print(f"No training_results_long.csv files found under {folder}")

    return written_files


written_files = generate_top5_training_csvs(root_folder)
print(f"Created {len(written_files)} top5_training.csv files")
```
