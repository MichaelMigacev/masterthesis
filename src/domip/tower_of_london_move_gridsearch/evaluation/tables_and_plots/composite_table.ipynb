```python
# This file contains code written by AI.

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

tables_folder = Path(r'c:\Users\Michael Migacev\Desktop\UNI\Wirklich_UNI\AICON\master-thesis-tol\master-thesis-tol\tower_of_london_plots\tables')

# Load and merge each split pair
splits_config = [
    {
        'name': '6_6_avg_success',
        'main': '6_6_avg_success.csv',
        'bfs': 'bfs_6_6_success_rate_validation.csv',
    },
    {
        'name': '6_6_extra_moves',
        'main': '6_6_extra_moves.csv',
        'bfs': 'bfs_6_6_extra_moves_validation.csv',
    },
    {
        'name': '12_12',
        'main': '12_12.csv',
        'bfs': 'bfs_12_12_validation.csv',
    },
]

composite_dfs = []
for config in splits_config:
    main_df = pd.read_csv(tables_folder / config['main'])
    bfs_df = pd.read_csv(tables_folder / config['bfs'])
    
    # Rename BFS columns to avoid conflicts
    bfs_df = bfs_df.rename(columns={'mean_tau': 'bfs_mean', 'std_tau': 'bfs_std'})
    
    # Drop alpha and beta from main_df if they exist
    main_df = main_df.drop(columns=['alpha', 'beta'], errors='ignore')
    
    # Merge on human_metric and participant_group
    merged = main_df.merge(
        bfs_df[['human_metric', 'participant_group', 'bfs_mean', 'bfs_std']],
        on=['human_metric', 'participant_group'],
        how='left'
    )
    
    # Add split identifier
    merged['split_type'] = config['name']
    composite_dfs.append(merged)

# Combine all splits
composite_df = pd.concat(composite_dfs, ignore_index=True)

# Reorder columns for readability
col_order = [
    'split_type',
    'human_metric',
    'participant_group',
    'train_mean',
    'train_std',
    'validation_mean',
    'validation_std',
    'bfs_mean',
    'bfs_std',
]
composite_df = composite_df[[c for c in col_order if c in composite_df.columns]]

print(f"Composite table shape: {composite_df.shape}")
print(f"Splits: {composite_df['split_type'].value_counts().to_dict()}")
composite_df
```

```python
output_name = 'composite_table.png'
output_path = tables_folder / output_name

render_df = composite_df.copy().fillna('')
fig_width = max(12, len(render_df.columns) * 1.5)
fig_height = max(4, 0.4 * len(render_df) + 1.8)

fig, ax = plt.subplots(figsize=(fig_width, fig_height))
ax.axis('off')

cell_text = []
for _, row in render_df.iterrows():
    formatted_row = []
    for value in row:
        if isinstance(value, float):
            formatted_row.append(f'{value:.3f}')
        else:
            formatted_row.append(str(value))
    cell_text.append(formatted_row)

table = ax.table(
    cellText=cell_text,
    colLabels=render_df.columns.tolist(),
    loc='center',
    cellLoc='center',
)
table.auto_set_font_size(False)
table.set_fontsize(8)
table.scale(1, 1.25)

plt.savefig(output_path, dpi=300, bbox_inches='tight')
plt.close(fig)

print(f"PNG saved to: {output_path}")
output_path
```
